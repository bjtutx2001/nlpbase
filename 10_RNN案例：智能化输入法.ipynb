# 智能输入法案例介绍


目标：根据用户已经输入的几个字，预测接下来可能会输入的字

项目结构：
- 项目目录
    - data：数据目录
        - processed：预处理后的数据
        - raw：源数据
    - logs：日志数据
    - models：保存训练好的模型参数
    - src：源码文件
        - config.py：超参数配置
        - dataset.py：自定义Dataset
        - eval.py：评估脚本
        - model.py：模型定义脚本
        - pre.py：模型预测脚本
        - process.py：模型预处理
        - tokenizer.py：自定义分词器
        - train.py：模型训练脚本

```python
#演示读取json文件
import pandas as pd
from json import loads, dumps
```

```python
#创建json文件
df = pd.DataFrame(
    {'a':[1,4,7],
     'b':[2,5,8],
     'c':[3,6,9]
    }
 )
df.index=['x','y','z']
df
```

```python
#生成json文件
df.to_json('./data/json1.json',orient='columns')
df.to_json('./data/json2.json',orient='index')
df.to_json('./data/json3.json',orient='split')
df.to_json('./data/json4.json',orient='records')
df.to_json('./data/json5.json',orient='records',lines=True)
```
