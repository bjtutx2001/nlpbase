# 演示LSTM API的使用

```python
from torch import nn
import torch
from unicodedata import bidirectional
```

```python
lstm=nn.LSTM(
    input_size=16,       #序列长队
    hidden_size=32,      #隐藏状态维度
    bidirectional=True,     #是否启用双向
    batch_first=True,       #批次大小
    bias=True,              #是否启用偏置
    num_layers=1,           #层数
    dropout=0.0,            #随机失活概率
    proj_size=0             #隐藏状态投影后的纬度，0表示不投影仪

)
```
