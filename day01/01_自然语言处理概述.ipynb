# 自然语言处理概述

自然语言处理（Natural Language Processing，NLP）是人工智能（AI）的一个重要分支，主要研究如何让计算机理解、分析、处理和生成人类使用的自然语言。
这里的自然语言指的是人类日常交流使用的语言，例如中文、英文、日文等，而不是 Python、Java 这样的编程语言。
NLP 的核心目标可以概括为两个方面：
- 自然语言理解（NLU）：让计算机理解文本表达的含义，例如情感分析、文本分类、问题理解等。
- 自然语言生成（NLG）：让计算机根据已有信息生成自然语言，例如机器翻译、文章生成、歌词生成和智能对话等。

自然语言处理包含多种典型任务：
- 文本分类
- 序列标注：对文本中的词打标签
- 文本生成：根据已有内容生成新的自然语言文本
- 信息抽取：从文本中提取出结构化的信息（给出 一段文本和 一个问题，从文本中抽取出答案，而不是生成答案）
- 文本转换：比如翻译和摘要生成

NLP演进历史：
- 规则系统阶段：查词典（人为规定），按照语法规则调整结构
- 统计方法阶段：比如n-gram(早起输入法使用)
- 机器学习阶段
- 深度学习阶段

词表（Vocabulary）：记录语料中有哪些不同的词，以及每个词对应的索引。
词袋模型（Bag of Words，BoW）：利用词表，将一段文本转换为一个数值向量，通常记录每个词在文本中出现的次数。词袋具有**无序性**，两句不一样的话词相同，词袋模型一样

在自然语言处理（NLP）中，Token（词元）是文本经过分词器（Tokenizer）切分后得到的基本处理单位。
